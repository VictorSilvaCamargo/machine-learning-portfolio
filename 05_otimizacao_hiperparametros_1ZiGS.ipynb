```python
# Importar as bibliotecas necessárias
import pandas as pd
from pyspark.sql import SparkSession
from pyspark.sql.functions import regexp_extract, col

# Criar a sessão do Spark
spark = SparkSession.builder.master("local").appName("Processar Excel").getOrCreate()

# Carregar o arquivo Excel com pandas, ignorando a primeira linha de cabeçalho
df_pandas = pd.read_excel('/content/SAI.xlsx', header=1, engine='openpyxl')

# Converter o pandas DataFrame para um PySpark DataFrame
df_spark = spark.createDataFrame(df_pandas)

# Verificar as primeiras linhas para garantir que as colunas foram carregadas corretamente
df_spark.show(5)

# Acessar diretamente a segunda coluna (índice 1)
df_with_parentheses = df_spark.withColumn(
    "extraido_parenteses",
    regexp_extract(col(df_spark.columns[1]), r"\((.*?)\)", 1)
)

# Coletar os valores extraídos da coluna "extraido_parenteses" em um vetor (lista)
resultado = df_with_parentheses.select("extraido_parenteses").rdd.flatMap(lambda x: x).collect()

# Criar uma lista para armazenar as palavras únicas
palavras_unicas = []

# Verificar se a palavra já foi adicionada, caso contrário, adicionar à lista
for item in resultado:
    if item not in palavras_unicas:
        palavras_unicas.append(item)

# Exibir as palavras únicas pulando linha
for item in palavras_unicas:
    print(item)

# Fechar a sessão do Spark
spark.stop()
```

```python
# Importar as bibliotecas necessárias
import pandas as pd
from pyspark.sql import SparkSession
from pyspark.sql.functions import regexp_extract, col

# Criar a sessão do Spark
spark = SparkSession.builder.master("local").appName("Processar Excel").getOrCreate()

# Carregar o arquivo Excel com pandas, ignorando a primeira linha de cabeçalho
df_pandas = pd.read_excel('/content/SAI.xlsx', header=1, engine='openpyxl')

# Converter o pandas DataFrame para um PySpark DataFrame
df_spark = spark.createDataFrame(df_pandas)

# Verificar as primeiras linhas para garantir que as colunas foram carregadas corretamente
df_spark.show(5)

# Acessar diretamente a segunda coluna (índice 1)
df_with_parentheses = df_spark.withColumn(
    "extraido_parenteses",
    regexp_extract(col(df_spark.columns[1]), r"\((.*?)\)", 1)
)

# Coletar os valores extraídos da coluna "extraido_parenteses" em um vetor (lista)
resultado = df_with_parentheses.select("extraido_parenteses").rdd.flatMap(lambda x: x).collect()

# Criar uma função para procurar a palavra exata e mostrar a linha
def encontrar_linha(palavra):
    linhas_encontradas = []
    for i, item in enumerate(resultado):
        if item == palavra:  # Compara a palavra exatamente, considerando maiúsculas e minúsculas
            linhas_encontradas.append(i + 2)  # +2 porque as linhas começam em 1 e estamos ignorando a primeira
    return linhas_encontradas

# Solicitar a entrada do usuário
palavra_busca = input("Digite a palavra exata para buscar dentro dos parênteses: ")

# Encontrar e exibir as linhas onde a palavra está presente
linhas = encontrar_linha(palavra_busca)

if linhas:
    print(f"A palavra '{palavra_busca}' foi encontrada nas seguintes linhas: {linhas}")
else:
    print(f"A palavra '{palavra_busca}' não foi encontrada em nenhum parêntese.")

# Fechar a sessão do Spark
spark.stop()
```

```python
import pandas as pd
from pyspark.sql import SparkSession
from pyspark.sql.functions import regexp_extract, col, when

# Criar a sessão do Spark
spark = SparkSession.builder.appName("Processar Excel").getOrCreate()

# Carregar o arquivo Excel com pandas
df_pandas = pd.read_excel('/content/CCO.xlsx', header=0, engine='openpyxl')

# Converter o pandas DataFrame para um PySpark DataFrame
df_spark = spark.createDataFrame(df_pandas)

# Dicionário de mapeamento
mapeamento = {
    "Alterar": "AL",
    "Excluir": "AL",
    "Pesquisar": "PE",
    "Listar": "PE",
    "PDF": "AL",
    "Inserir": "AL",
    "Executar": "AL",
    "Selecionar": "PE",
    "Bloquear": "AL",
    "Excel": "AL",
    "Filtrar": "PE",
    "Enviar": "AL",
    "Desbloquear": "AL",
    "Emitir": "AL",
    "Vincular": "AL",
    "Cadastrar": "AL",
    "Importar": "AL",
    "Consultar": "PE",
    "Detalhar": "PE",
    "Simular": "PE",
    "Gerar": "AL",
    "Reabrir": "AL",
    "Inscrever": "AL",
    "Exportar": "AL",
    "TXT": "AL",
    "Incluir": "AL",
    "Conveniar": "PE",
    "Cancelar": 'PE',
    "Validar": "PE",
    "Calcular": "PE"
}

# Nome da terceira coluna (ajuste se necessário)
nome_terceira_coluna = df_spark.columns[2]

# Extrair texto entre parênteses da segunda coluna e aplicar o mapeamento
df_with_mapping = df_spark.withColumn(
    nome_terceira_coluna,
    when(
        col(df_spark.columns[1]).rlike(r"\((.*?)\)"),  # Verifica se há parênteses
        regexp_extract(col(df_spark.columns[1]), r"\((.*?)\)", 1)  # Extrai texto entre parênteses
    ).otherwise(col(df_spark.columns[1]))  # Mantém o valor original se não houver parênteses
).withColumn(
    nome_terceira_coluna,
    when(
        col(nome_terceira_coluna).isin(mapeamento.keys()),  # Verifica se o valor está no mapeamento
        col(nome_terceira_coluna).rlike(r"\((.*?)\)")  # Mantém o valor original se não houver parênteses
    ).otherwise(col(nome_terceira_coluna)) # Mantém o valor original se não houver parênteses
).withColumn(
    nome_terceira_coluna,
    when(col(nome_terceira_coluna).isin(mapeamento.keys()), mapeamento.get(col(nome_terceira_coluna).getItem(0)))
         .otherwise(col(nome_terceira_coluna)))


# Converter de volta para pandas DataFrame
df_resultado_pandas = df_with_mapping.toPandas()

# Salvar em um novo arquivo Excel
novo_arquivo = "/content/resultado_processado.xlsx"
df_resultado_pandas.to_excel(novo_arquivo, index=False, engine='openpyxl')

print(f"O arquivo foi salvo em: {novo_arquivo}")

spark.stop()
```
