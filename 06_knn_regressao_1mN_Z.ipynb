```python
# incluir as funções e métodos necessários

from sklearn.neighbors import KNeighborsRegressor
from sklearn import model_selection
from sklearn.metrics import mean_squared_error, mean_absolute_error
from sklearn.datasets import load_diabetes
from sklearn import tree
```

```python
# load dataset
X,y = load_diabetes(return_X_y=True)
```

```python
#Model declaration

#reg = KNeighborsRegressor(n_neighbors=5, weights='distance')
regKnn = KNeighborsRegressor(n_neighbors=5, weights='distance')

regDT = tree.DecisionTreeRegressor(criterion='absolute_error',max_depth=10 ,random_state=46)

#CrossValidation (5 folds)
#result=model_selection.cross_val_score(reg,X,y,cv=5)
resultKnn=model_selection.cross_val_score(regKnn,X,y,cv=5)
print('KNN Coefficient of determination R2: ', resultKnn.mean())

resultDT=model_selection.cross_val_score(regDT,X,y,cv=5)
print('DT Coefficient of determination R2: ', resultDT.mean())

#Model Prediction
#y_pred = model_selection.cross_val_predict(reg,X,y,cv=5)
#print("Absolute Error: ", mean_absolute_error(y,y_pred))

y_predKnn = model_selection.cross_val_predict(regKnn,X,y,cv=5)
print("Knn Absolute Error: ", mean_absolute_error(y,y_predKnn))

y_predDT = model_selection.cross_val_predict(regDT,X,y,cv=5)
print("DT Absolute Error: ", mean_absolute_error(y,y_predDT))

```
